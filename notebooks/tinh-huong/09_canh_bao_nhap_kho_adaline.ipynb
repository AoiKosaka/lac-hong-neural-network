## Bối cảnh

Một cửa hàng muốn cảnh báo mặt hàng nào cần được nhập thêm trong kỳ kế tiếp.

Feature có thể cân nhắc: số lượng tồn kho, lượng bán trung bình, số ngày chờ hàng, số đơn đang chờ (chuẩn hoá).

## Vì sao dùng Adaline thay Perceptron?

Thay vì chỉ 0/1 (chưa cần / cần nhập thêm), hãy coi target là **số lượng nên nhập thêm** hoặc **điểm mức độ khẩn cấp cần nhập hàng** trong [0, 1].

Nhắc lại công thức Adaline / Widrow–Hoff:

$$
\text{net} = \mathbf{w}^\top \mathbf{x} + b, \qquad e = t - \text{net}
$$

$$
\mathbf{w} \leftarrow \mathbf{w} + \eta e \mathbf{x}, \qquad b \leftarrow b + \eta e
$$

Dự đoán số lượng cần nhập (một số thực) trực tiếp hữu ích hơn cho việc lập đơn hàng so với chỉ biết cần nhập hay không.
